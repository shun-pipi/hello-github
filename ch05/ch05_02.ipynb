# 05-02 複雑なデータ構造

## インターネットからデータ構造を取得する

```python
# 5・2・1
from urllib.request import urlopen
url = "https://www.jma.go.jp/bosai/amedas/const/amedastable.json"
json_str = urlopen(url).read()
```

```python
# 5・2・2
amedas = eval(json_str)
```

```python
# 5・2・3
amedas.keys()
```

```python
# 5・2・4
amedas["11001"]
```

## 辞書からデータを抜き出す

```python
# 5・2・5
names = []   # 漢字名を追加するリストを初期化
for point_id in amedas.keys():
    name = amedas[point_id]["kjName"]
    names.append(name)
names    # 結果を表示
```

```python
# 5・2・6
lats = []   # 緯度を追加するリストを初期化
lons = []   # 経度を追加するリストを初期化
for point_id in amedas.keys():
    # 緯度と経度の度と分を取り出す
    lat_d = amedas[point_id]["lat"][0]
    lat_m = amedas[point_id]["lat"][1]
    lon_d = amedas[point_id]["lon"][0]
    lon_m = amedas[point_id]["lon"][1]
    # 分を度に追加
    lat = lat_d+lat_m/60
    lon = lon_d+lon_m/60

    # 緯度と経度をリストに追加
    lats.append(lat)
    lons.append(lon)
```

## データをグラフに出力する

```python
# 5・2・7
import matplotlib.pyplot as plt
plt.scatter(lons, lats, s=1)
```
